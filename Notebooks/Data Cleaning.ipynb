# GSE63060

## Loading Dataset

```python
file_path = "/content/drive/MyDrive/Colab Notebooks/Personal/brain_disease_prediction/GEO/Alzheimer/blood tissue/GSE63060_family.soft.gz"

gse_60 = GEOProcessor(file_path)
```

```python
gse_60.load_data()
```

## Exploring the dataset

```python
gse_60.show_info()
```

```python
print(f'N° of patients:{len(gse_60.get_patient_ids())}')
print(gse_60.get_patient_ids())
```

```python
print(gse_60.get_table_columns())
```

```python
gse_60.show_patient_data("GSM1539080")
```

```python
gse_60.show_patient_metadata("GSM1539080")
```

## Building the Expression DataFrames

```python
dfs_gse_60 = gse_60.build_data_frames()
```

```python
df_gse_60 = dfs_gse_60["value"]
df_gse_60
```

## Building the metadata DataFrame

```python
gse_60.get_patient_labels("GSM1539080")
```

```python
metadata_gse_60 = gse_60.build_gse_metadata().set_index('patient_id')
metadata_gse_60
```

## Mapping the Probe names with their corresponding gene name

```python
gse_60.get_platforms()
```

```python
gse_60.get_platform_columns('GPL6947')
```

```python
gse_60.show_platform_columns_data('GPL6947', columns=['ID', 'ILMN_Gene', 'Entrez_Gene_ID', 'Symbol'])
```

Here we can notice that the ID are of course the same as the gene expression dataframe which represent the probe id's, the Symbol column to be the official gene name detected by the probe

```python
probe_to_gene = gse_60.build_probe_mapping('GPL6947', probe_col='ID', gene_col='Symbol')

temp_df = pd.DataFrame(probe_to_gene.items(), columns=['probe_id', 'gene_name'])
temp_df
```

```python
final_df_60 = gse_60.replace_probe_names(df_gse_60, 'GPL6947', probe_col='ID', gene_col='Symbol')
```

```python
final_df_60
```

## Removing unwanted and null Genes

```python
final_df_60.shape
```

### Null gene names

```python
final_df_60 = final_df_60[ final_df_60.index.notna() ]
```

```python
final_df_60.shape
```

### Empty gene names

```python
final_df_60 = final_df_60[ final_df_60.index != "" ]
```

```python
final_df_60.shape
```

### Removing Affymetrix control probes

```python
final_df_60 = final_df_60[~final_df_60.index.str.startswith("AFFX")]
```

```python
final_df_60.shape
```

### Duplicated genes

```python
print(f'N° of duplicated genes: {final_df_60.index.duplicated().sum()}')
```

```python
final_df_60.groupby(final_df_60.index).size().sort_values(ascending=False).head(20).plot(kind='bar')
```

```python
final_df_60 = final_df_60.groupby(final_df_60.index).mean()
```

```python
final_df_60.groupby(final_df_60.index).size().sort_values(ascending=False).head(20).plot(kind='bar')
```

```python
final_df_60.shape
```

## EDA

```python
metadata_gse_60
```

* **MCI (Mild Cognitive Impairment)**: This represents an intermediate stage between expected, normal cognitive aging and actual dementia. Individuals with MCI typically have minor but noticeable problems with memory, language, or thinking.
* **AD (Alzheimer's Disease)**: This is a progressive, neurodegenerative brain disorder. In the context of MCI studies, AD indicates patients who have advanced beyond MCI into clinical dementia, where their ability to function independently is impaired.
* **CTL (Control)**: This refers to the Healthy Control group. In clinical trials and research, these are neurologically normal individuals of similar age who serve as a baseline to compare cognitive test scores, brain scans, or physical markers against MCI and AD patient

```python
status_counts = GSE63060["status"].value_counts()

# Create figure
plt.figure(figsize=(8, 6))

# Bar plot
bars = plt.bar(
    status_counts.index,
    status_counts.values,
    width=0.6
)

# Add values on top
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 2,
        f"{int(height)}",
        ha='center',
        fontsize=11,
        fontweight='bold'
    )

# Labels and title
plt.title(
    "Distribution of Diagnostic Classes",
    fontsize=16,
    fontweight='bold',
    pad=15
)

plt.xlabel("Status", fontsize=13)
plt.ylabel("Number of Samples", fontsize=13)

# Grid
plt.grid(axis='y', alpha=0.25)

# Clean top/right borders
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Improve ticks
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

plt.tight_layout()
plt.show()
```

```python
metadata_gse_60["age"].astype(int).plot(kind='box')
```

## Adding the patients diagnosis to the expression Dataframe

```python
final_df_60 = final_df_60.T
```

```python
final_df_60.head(1)
```

```python
metadata_gse_60.head(1)
```

```python
metadata_gse_60 = metadata_gse_60.reindex(final_df_60.index)
```

```python
metadata_gse_60
```

```python
final_df_60['status'] = metadata_gse_60['status']
```

```python
final_df_60['status']
```

```python
all(metadata_gse_60['status'] == final_df_60['status'])
```

## Checking null Gene expression values

```python
final_df_60.isna().sum().sort_values(ascending=False)
```

## Exporting the dataset

```python
final_df_60.to_csv("/content/drive/MyDrive/Colab Notebooks/Personal/brain_disease_prediction/GEO/Alzheimer/blood tissue/GSE63060_expression.csv")
```
